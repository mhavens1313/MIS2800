```python
%%sql
-- Create the table
CREATE TABLE products (
    product_id INTEGER PRIMARY KEY,
    name TEXT,
    price REAL
) STRICT;

-- Insert 3 items (The instructions require at least 3, the example only showed 1)
INSERT INTO products (name, price) VALUES ('Superbowl Jersey', 120.00);
INSERT INTO products (name, price) VALUES ('Basketball', 25.50);
INSERT INTO products (name, price) VALUES ('Running Shoes', 89.99);
```

```python
%%sql
-- Show all products that cost more than $50
SELECT * FROM products WHERE price > 50;
```

```python
%%sql
-- Update: Change the price of the Jersey
UPDATE products SET price = 99.99 WHERE name = 'Superbowl Jersey';

-- Delete: Remove the item with ID 1 (which is usually the first item you inserted)
DELETE FROM products WHERE product_id = 1;

-- Verify your changes (Optional but smart)
SELECT * FROM products;
```
