## YOUR ANSWERS TO PART 2 BELOW

Question 1: Column storage is something that analysts are more concerned about because they want to see one piece of information out of millions of pieces of data, rather than reading all the details of one single transaction. The database only needs to read the exact information that it needs because of column storage instead of row storage, which can save a huge amount of time when running large reports.

Question 2: Think of the "Leader Node" as a manager and the "Compute Nodes" as a team of workers. When you ask for a massive report, the manager breaks the giant task into smaller pieces and hands them out to the workers. Because the whole team is crunching their specific slice of the data at the exact same time, the final report gets stitched together in a fraction of the time it would take one computer working alone.
